```python
df = pd.read_csv("C:/Office 2019/NLP/Fake_News_Detection_System/Dataset/dataset.csv")

df1 = pd.DataFrame({'text':df['Real News'],"label":"REAL"})
df2 = pd.DataFrame({'text':df['Fake News'],"label":"FAKE"})
df3 = pd.DataFrame({'text':df['AI-generated Fake News'],"label":"FAKE"})

df = pd.concat([df1,df2,df3],ignore_index=True,)
df
```

```python
df['text'] = df['text'].apply(lambda x: str(x).lower())
df
```

```python
import string
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
def remove_putch_stopword_number(txt):
    txt = re.sub(r'\d+','',txt)
    txt = re.sub(r'["“”\'‘’]', '', txt)
    words = word_tokenize(txt)

    stop_words = set(stopwords.words('english'))
    clean_words = [word for word in words if word not in string.punctuation and word not in stop_words and not word.isdigit()]

    return ' '.join(clean_words)

df["text"] = df["text"].apply(remove_putch_stopword_number)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'],df['label'], test_size=0.33, random_state=42)
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
# 1 -> True and 0 -> False
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

log_model = LogisticRegression(max_iter=10000)
log_model.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
log_pred = log_model.predict(X_test)
accuracy_score(y_test,log_pred)
```
